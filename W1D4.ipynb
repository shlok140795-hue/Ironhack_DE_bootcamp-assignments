```python
#Practice Questions
#Data Understanding
#Load the dataset into a DataFrame called df. 
df = pd.read_csv("nba.csv")
```

```python
#Display the first 5 rows. 
print(df.head(5))
```

```python
df = df.dropna(how="all") #drops rows which are completely empty
#Check the number of rows and columns. 
print(df.shape)
```

```python
#List all column names. 
print(df.columns)
```

```python
print(df.info())
```

```python
#Data Cleaning 
#Convert the Salary column to numeric (handle errors). 
df["Salary"] = pd.to_numeric(df["Salary"], errors="coerce")
df["Salary"]
```

```python
#Count how many missing (NaN) values are in each column
print(df.isna().sum())
```

```python
#Drop rows where Salary is missing. 
df = df.dropna(subset = ["Salary"])
```

```python
#Fill missing College values with "Unknown". 
df = df.fillna({"College": "Unknown"})
```

```python
#Filtering Data
#Show all players from Boston Celtics. 
boston_players = df[df["Team"] == "Boston Celtics"]
print(boston_players)
```

```python
#Show players whose Age is greater than 30. 
players_above_30 = df[df["Age"] >30]
print(players_above_30)
```

```python
#Filter players who play as PG (Point Guard)
players_pos_as_PG = df[df["Position"] == "PG"]
print(players_pos_as_PG)
```

```python
#Find players with Salary greater than 5,000,000.
salary_above_5M = df[df["Salary"] > 5000000]
print(f"salary_above_5: {salary_above_5M}")
```

```python
# Sorting
#Sort players by Salary (highest first). 
salary_sorted = df.sort_values(by = "Salary", ascending = False)
print(f"salary_sorted: {salary_sorted}")
```

```python
#Sort players by Age (youngest first). 
sort_age = df.sort_values(by= "Age")
print(f"sorted_age: {sort_age}")
```

```python
#Find the average salary per team.
avg_salary_per_team = df.groupby("Team")["Salary"].mean().round(2)
print(f"avg_salary_per_team: {avg_salary_per_team}")
```

```python
#Count number of players in each team. 
count_per_team = df.groupby("Team").size()
print(f"count_per_team: {count_per_team}")
```

```python
#Find the maximum salary in each team. 
max_salary_per_team = df.groupby("Team")["Salary"].max() 
print(f"max_salary_per_team: {max_salary_per_team}")
```

```python
#Aggregations
#What is the average age of players? 
average_age = df["Age"].mean().round(2)
print(f"average_age: {average_age}")

#What is the maximum salary in the dataset? 
max_salary = df["Salary"].max()
print(f"max_salary: {max_salary}")

#What is the minimum weight? 
min_weight = df["Weight"].min()
print(f"min_weight: {min_weight}")
```

```python
#Column Operations
#Create a new column Age_in_5_years. 
df["Age_in_5_years"] = df["Age"] + 5
print(df[["Age", "Age_in_5_years"]])
#Create a new column Salary_in_Millions (Salary / 1000000). 
df["Salary_in_Millions"] = df["Salary"] / 1000000
print(df[["Salary", "Salary_in_Millions"]])
```

```python
#Bonus 
#Which team has the highest total salary payout? 
highest_total_salary_per_team = df.groupby("Team")["Salary"].sum().max()
print(f"highest_total_salary_per_team: {highest_total_salary_per_team}")
team_highest_total_salary = df.groupby("Team")["Salary"].sum().idxmax()  #.idxmax() gives the name/index belonging to the highest value
print(f"team_highest_total_salary: {team_highest_total_salary}")
#Which position has the highest average salary?
highest_avg_salary_per_pos = df.groupby("Position")["Salary"].mean().round(2).max()
print(f"highest_avg_salary_per_pos: {highest_avg_salary_per_pos}")
pos_highest_avg_salary = df.groupby("Position")["Salary"].mean().idxmax()
print(f"pos_highest_avg_salary: {pos_highest_avg_salary}")
```

```python
highest_avg_salary_per_pos = df.groupby("Position")["Salary"].mean().round(2).sort_values(ascending=False)
print(f"highest_avg_salary_per_pos: {highest_avg_salary_per_pos}")
```

```python
team_highest_total_salary = df.groupby("Team")["Salary"].sum().sort_values(ascending=False)
print(f"team_highest_total_salary: {team_highest_total_salary}")
```
