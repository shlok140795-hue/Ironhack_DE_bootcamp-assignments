```python
def area_square(side):
    return side * side

def area_circle(radius):
    return 3.14 * radius ** 2

def area_rectangle(length, breadth):
    return length * breadth

def area(choice):
    if choice == 1:
        side = float(input("Enter the side: "))
        return area_square(side)

    elif choice == 2:
        radius = float(input("Enter the radius: "))
        return area_circle(radius)

    elif choice == 3:
        length = float(input("Enter the length: "))
        breadth = float(input("Enter the breadth: "))
        return area_rectangle(length, breadth)

    else:
        return "Invalid choice"


user_choice = int(input("Choose 1. Area of Square 2. Area of circle  3. Area of Rectangle "))

result = area(user_choice)
result
```

```python
#Write a function calculate_sum(n) that uses a loop to calculate and return the sum of numbers from 1 to n.
def calculate_sum(n):
    total=0
    for i in range(1,n+1):
     total+=i
    return total
    
num = int(input("Enter a number")) 
sum_of_numbers = calculate_sum(num)
print(f"Sum of numbers from 1 to given number is {sum_of_numbers}")
```

```python
# Write a function count_greater(numbers, value) that accepts a list of numbers and a value, and returns how many numbers in the list are greater than that value.
def count_greater(numbers, value):
    count = 0
    for i in numbers:
     if i>value:
         count+= 1
    return count
num_list = [int(x) for x in input("Enter numbers separated by spaces: ").split()]
print(num_list)
comparing_value = float(input("Enter the value to be compared"))
count_of_greater_numbers = count_greater(num_list, comparing_value)
print(f"Count of numbers greater than value in the list are {count_of_greater_numbers}")
```

```python
#Write a function called analyze_numbers(numbers) that accepts a list of numbers and calculates:
def analyze_numbers(numbers):
    total_values = 0
    largest_value = numbers[0]
    smallest_value = numbers[0]
    even_count = 0
    odd_count = 0
    total = 0
    for i in numbers:
        #Total number of values 
        total_values+=1
        
        #Sum of the values 
        total+=i
    
        #Largest value 
        if i>largest_value:
            largest_value=i
            
        #Smallest value 
        if i<smallest_value:
            smallest_value=i
        #Number of even values 
        if i%2 == 0:
           even_count += 1
        #Number of odd values
        else:odd_count+= 1
    #Average 
    avrg = total/total_values
    print(f"Total number of values: {total_values}")
    print(f"Sum of the values: {total}")
    print(f"Average: {avrg:.2f}")
    print(f"Largest value: {largest_value}")
    print(f"Smallest value: {smallest_value}")
    print(f"Number of even values: {even_count}")
    print(f"Number of odd values: {odd_count}")


num_list = [int(x) for x in input("Enter numbers separated by spaces: ").split()]
analyze_numbers(num_list)
```

```python
#with built-in functions 
def analyze_numbers(numbers):
    total_values = len(numbers)
    total = sum(numbers)
    average = total / total_values
    largest_value = max(numbers)
    smallest_value = min(numbers)
    even_count = sum(1 for i in numbers if i % 2 == 0)
    odd_count = sum(1 for i in numbers if i % 2 != 0)

    print(f"Total number of values: {total_values}")
    print(f"Sum of the values: {total}")
    print(f"Average: {average:.2f}")
    print(f"Largest value: {largest_value}")
    print(f"Smallest value: {smallest_value}")
    print(f"Number of even values: {even_count}")
    print(f"Number of odd values: {odd_count}")

num_list = [int(x) for x in input("Enter numbers separated by spaces: ").split()]
analyze_numbers(num_list)
```
