```python

from pyspark.sql.functions import col

# Select the required columns from the dataset

twitter_df = df.select(
    col("tweet_id"),
    col("created_at"),
    col("text"),
    col("lang"),
    col("user.username").alias("username"),
    col("user.country").alias("country"),
    col("user.followers_count").alias("followers_count"),
    col("metrics.likes").alias("likes"),
    col("metrics.retweets").alias("retweets"),
    col("hashtags")
)

# Display cleaned dataset

display(twitter_df)
```

```python
#Question 1: Top 5 users with the highest number of tweets
from pyspark.sql.functions import count, col
top_5_users = (
    twitter_df
    .groupBy("username")
    .agg(count("tweet_id").alias("total_tweets"))
    .orderBy(col("total_tweets").desc())
    .limit(5)
)

display(top_5_users)
```

```python
#Most commonly used words in all the tweets, excluding stopwords
from pyspark.sql.functions import explode, split, lower, regexp_replace, trim, col, collect_list
from pyspark.ml.feature import StopWordsRemover

# Step 1: Clean tweet text
cleaned_tweets = twitter_df.select(
    lower(
        regexp_replace(col("text"), "[^a-zA-Z\\s]", "")
    ).alias("clean_text")
)

# Step 2: Split tweets into individual words
words_df = cleaned_tweets.select(
    explode(split(col("clean_text"), "\\s+")).alias("word")  #"\\s+" A whitespace character (space, tab, newline) with one or more occurences
)

# Step 3: Remove empty words
words_df = words_df.filter(col("word") != "")

# Step 4: Remove stopwords
remover = StopWordsRemover(
    inputCol="words",
    outputCol="filtered_words"
)

# Convert individual words into arrays for StopWordsRemover
word_arrays = words_df.groupBy().agg(
    collect_list("word").alias("words")
)

filtered = remover.transform(word_arrays)

# Step 5: Count word frequency
final_words = filtered.select(
    explode(col("filtered_words")).alias("word")
)

word_frequency = (
    final_words
    .groupBy("word")
    .agg(count("*").alias("frequency"))
    .orderBy(col("frequency").desc())
    .limit(20)
)

display(word_frequency)
```

```python
#Which country has the highest number of tweets?

from pyspark.sql.functions import count, col

country_tweets = (
    twitter_df
    .filter(
        col("country").isNotNull() &
        (col("country") != "")
    )
    .groupBy("country")
    .agg(count("tweet_id").alias("total_tweets"))
    .orderBy(col("total_tweets").desc())
)

display(country_tweets.limit(10))
```
